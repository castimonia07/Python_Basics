# Importing Modules in Python: Modules and Packages

In Python,modules and packages help organize and reuse code.Here's a comprehensive guide on how to import them.

```python
import math  # its just like a module which have many functions

math.sqrt(16)  # square root of 16
```

```python
from math import sqrt,pi # importing only the sqrt function from math module
print(sqrt(25)) 
print(pi)  # printing the value of pi
```

```python
import numpy as np # importing numpy module as np
# use numpy to create an array 
# as here is an alias np is used to call the numpy module

np.array([1,2,3,4,5,6]) # creating an array using numpy
```

```python
from math import * # used to import all the functions from math module

print(sqrt(16))
print(pi)
```

```python
from package.maths import addition
addition(2,3)
```
