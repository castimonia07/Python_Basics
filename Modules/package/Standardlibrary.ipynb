# Data Serialization and Deserialization
it is used to covert a json object to python object and vice versa.

```python
import json 
data={'name':'Rishabh',
      'age':20,
      'city':'Varanasi'}

json_str=json.dumps(data)
print(json_str) # it will convert the python object to json string

print(type(json_str)) # it will print the type of json_str which is str

parsed_data=json.loads(json_str) # it will convert the jsonstring to python object
print(parsed_data)
print(type(parsed_data)) # it will print the type of parsed_data which is dict
```

```python
# csv module

import csv
with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age','city'])
    writer.writerow(['Rishabh',20,'Varanasi'])


with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
        
''' what it is doing ki ye ek csv file create kar raha hai jisme 2 rows
hai aur phir us csv file ko read kar raha hai aur print kar raha hai. 
'''

```

```python
# datetime module

from datetime import datetime,timedelta
now=datetime.now()
print(now) # it will print the current date and time

yesterday=now-timedelta(days=1) # it will print the date and time of yesterday
print(yesterday)
```

```python
# time module
''' it is used to work with time.'''
import time
print(time.time()) # it will print the current time in seconds since epoch
time.sleep(2) # it will pause the execution for 2 seconds

print(time.time()) # it will print the current time in seconds since epoch after 2 seconds
```

```python
# Regular expression module
import re

# it helps you to match a pattern in a string

pattern=r'\d+'  # here \d+ means one or more digits
text='My phone number  65416546 is 1234567890' # ek baar mil gya fir nhi check krta hai ye
match=re.search(pattern,text)
print(match.group()) # it will print the matched pattern which is 1234567890
```
