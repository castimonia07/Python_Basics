# Abstraction
Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object.This helps in reducing the programming complexity and efforts.

```python
from abc import ABC,abstractmethod

## Abstract base Class
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for Driving")

    @abstractmethod  # it is always a complete empty method
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car=Car()
operate_vehicle(car)

'''Ishme hua kya ki keval feature ko define kiya gaya hai aur uska 
implementation child class me kiya gaya hai. It hides the implementation
details from the user and only shows the functionality to the user.'''

```
