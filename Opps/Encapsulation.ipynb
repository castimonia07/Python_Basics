# Encapsulation And Abstraction

Encapsulation is the concept of wrapping data(variables) adn methods(functions) together as a single unit.It restricts direct acess to some of the object's component, which is a means of preventing accidental interference and misue of the data

```python
## Encapsulation with Getter and Setter Methods
## Acsess Methods: Public, private, Protected

class Person:
    def __init__(self,name,age):
        self.name=name  #public variable
        self.age=age  #public variable

def get_name(person):
    return person.name

person=Person("Rishabh",20)
# print(person.name)

get_name(person)

'''if i want ki varibales har jagah directlty acess na ho to us case 
me we make in use of private and protected variables and we can access
them through getter and setter methods.'''
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.__name=name  #private variable  __ cannot be accessed outside the class
        self.__age=age  #private variable __
        self.gender=gender #public variable

def get_name(person):
    return person.__name

person=Person("Rishabh",20,"Male")
# get_name(person)   # aceess nhi hoga 
```

```python
person=Person("Rishabh",20,"Male")
person = Person("Rishabh", 20, "Male")

print(vars(person))   # {'_Person__name': 'Rishabh', '_Person__age': 20, 'gender': 'Male'}
print(dir(person))    # Includes inherited and private attributes

```

```python
''' protected variable can be accessed outside the class but it is not
recommended to do so. It is recommended to use getter and setter methods
to access protected variables. It can only be accessed by the class and its subclassesor derieved class'''

class Person:
    def __init__(self,name,age,gender):
        self._name=name  #protected variable  _ can be accessed outside the class but it is not recommended to do so.
        self._age=age  #protected variable _
        self.gender=gender #public variable

class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

employee=Employee("Rishabh",20,"Male")
print(employee._name)  # it will print the name because it is protected variable and it
```

```python
# Accessing Using Getter and Setter Methods

class Person:
    def __init__(self,name,age):
        self.__name=name #private variable or modifier
        self.__age=age #private variable or modifier

    # getter method for name
    def get_name(self):   # it is used to retrieve the value of private variable __name
        return self.__name

    #setter method for name
    def set_name(self,name):  # it is used to set the value of private variable __name
        self.__name=name

    # getter method for age
    def get_age(self):
        return self.__age

    def set_age(self,age):  # it is used to set the value of private variable __age
        if age<0:
            print("Age cannot be negative")
        else:
            self.__age=age

person=Person("Rishabh",20)

# Access and Modify private Variables using Getter and Setter Methods

print(person.get_name())  # it will return the name of the person
print(person.get_age())  # it will return the age of the person

person.set_age(21) # it will set the age of the person to 21
print(person.get_age())  # it will return the age of the person

person.set_age(-5) # it will print the statement "Age cannot be negative" because age cannot be negative
```
