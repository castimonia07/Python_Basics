# Magic Methods

Also known as Dunder method(double underscore method).
They are the special methods that start and end with double underscores.These methods enable you to define the behavious of objects for built-in operations, such as arithmetic operations,comparisons etc,

```python
'''
__init__ : Intializes a new instance of a class.
__str__ : Returns a string represntation of the object.
__repr__ : return an unofficial string representation of the object.
__len__ : Returns the length of the object.
__getitem__ : Returns the value of the specified index.Get an item from container.
__setitem__ : Sets the value of the specified index.Set an item in container.
__delitem__ : Deletes the value of the specified index.Delete an item from container.
__contains__ : Returns True if the specified item is in the container, otherwise False.
__iter__ : Returns an iterator object for the container.
__next__ : Returns the next item from the iterator.
'''
```

```python
## Baic Magic Methods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

person=Person("Rishabh",20)
print(person)  # here deflaut string representation of the object will be printed which is not useful
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):  
        return f"{self.name},{self.age} years old"

    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"


person=Person("Rishabh",20)
print(person) # Now the string representation of the object will be printed which is useful

print(repr(person)) # Person(name=Rishabh,age=20)
```
