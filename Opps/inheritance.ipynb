# Inheritance in Python
allows a class to inherit attributes and methods from anther class.

```python
## Inheritance (Single Inheritance)
## Parent class
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")

    
```

```python
car1=Car(4,4,"Petrol")
# print(car1.drive())   # none dega becoause the drive method is not retruning anything

car1.drive()  # it will print the statement in the drive method of Car class
```

```python
# Child class
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_self_driving):
        super().__init__(windows,doors,enginetype)   
        self.is_self_driving = is_self_driving
        ''' super() is used to call the constructor of the parent class 
            means we have alredy declared the attributes in the parent calss 
            and we called it here '''

    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_self_driving}")
        
    
```

```python
tesla1=Tesla(4,5,"electric",True)
tesla1.selfdriving()
```

## Multiple Inheritace

when a class inherits from more than one base class



```python
# Base Class 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print("Subclass must implement this method")

# Base Class 2
class Pet:
    def __init__(self,owner):
        self.owner=owner

# Derived Class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)  # calling the constructor of Animal class
        Pet.__init__(self,owner)    # calling the constructor of Pet class

    def speak(self):
        return f"{self.name} says woof!"


# Create an objec
dog=Dog("Buddy","Rishabh")
print(dog.speak()) 
print(f"{dog.name} is owned by {dog.owner}")
```
