# Polymorphism
Same Name Multiple forms. Achieved through Method overriding and interfaces

# Method Overriding 
Method Overriding aloows a chlid class to provide a specific implementation of a method that is alredy defined in its parent class.

```python
## Base Class
class Animal:
    def speak(self):
        return "Sound of Animal"

# Derieved Class 1
class Dog(Animal):
    def speak(self):
        return "Woof!"   # same method name as in base class but different implementation

# Derieved Class 2
class Cat(Animal):
    def speak(self):
        return "Meow!"  # same method name as in base class but different implementation


## Function that demonstrates polymorphism
def animal_sound(animal):
    print(animal.speak()) # it will call the speak method of the class of the object passed as argument

dog=Dog()
cat=Cat()

print(dog.speak())  # it will call the speak method of Dog class
print(cat.speak())  # it will call the speak method of Cat class
```

```python
## Polymorphism with functions and methods

##Base class
class Shape:
    def area(self):
        return "The area of the figure"

## Derieved class 1
class rectangle(Shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    def area(self):
        return self.length*self.breadth  # same method name as in base class but different implementation

# Derieved class 2
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius # same method name as in base class but different implementation

## Function that demonstarates polymorphism

def print_area(shape):
    print(f"the area is:{shape.area()}") # it will call the area method of the class of the object passed as argument

rectangle1=rectangle(4,5)
circle1=Circle(3)

print_area(rectangle1)
print_area(circle1)
```

# Polymorphism With Abstract Base Class(Interfaces)

It is used to define common methods for a group of related objects.

```python
from abc import ABC,abstractmethod

## Define an abstract class  
''' Abstract class is a complete empty class that contains one or more abstract methods.
An abstract method is a method that is declared, but contains no implementation. 
Abstract classes cannot be instantiated, and require 
subclasses to provide implementations for the abstract methods.
'''
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

## Derieved class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine strated"

# Derieved class 2
class Bike(Vehicle):
    def start_engine(self):
        return "Bike engine started"

# Function that demonstrates polymorphism
def start_vehicle(vehicle):
    print(vehicle.start_engine()) # it will call the start_engine method of the class of the object passed as argument

# create objects of Car and Bike classes
car = Car()
bike = Bike()

start_vehicle(car)  # it will call the start_engine method of Car class
start_vehicle(bike)  # it will call the start_engine method of Bike class
```
