# Class and Objects

A class is a blue print for creating objects.It has its own attributes and methods which can be used by its objects

```python
class Car:
    pass

audi=Car()
bmw=Car()

print(type(audi))
```

```python
audi.window=4
print(audi.window) # window is an attribute of audi object

'''This is not a good way to define attributes of a class.
We should define the attributes in the class itself.'''

tata=Car()
tata.door=4
print(tata.door) # door is an attribute of tata object
```

```python
# Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):   # self or (anything inpace of self) is a reference to the current instance of the class
        self.name=name  #instance variable
        self.age=age

# object of the class
dog1=Dog("Tommy",5)  # it will call the constructor and assign the values to the instance variables
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2=Dog("Rocky",3)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
# INSTANCE METHODS
'''It is a function which is defined inside the class and it is used to perform 
some operations on the instance variables of the class.'''

class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):  # instance method
        print(f"{self.name} says Woof Woof")

dog1=Dog("Buddy",3)
dog1.bark() # calling the instance method

dog2=Dog("Lucy",5)
dog2.bark() # calling the instance method
```

```python
## Modeling a Bank Account 

# Define a class for bank Acoount

class BankAccount:
    def __init__(self,owner,balance=0):  # constructor
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):  # instance method
        self.balance+=amount
        print(f"{amount} is deposited.New Balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient Balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn. New Balance is {self.balance}")

    def get_balance(self):
        return self.balance

    # Creating an object
account=BankAccount("Rishabh",5000)
print(account.balance)  # it will give the initial balance of the account

account.deposit(100)
```

```python
# call instance method
account.deposit(500)  # it will deposit the amount in the account and update the balance
```

```python
print(account.get_balance())  # it will return the current balance of the account
```
