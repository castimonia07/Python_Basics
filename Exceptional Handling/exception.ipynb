# Exceptional Handling

Exceptional handling in python allows you to handle errors gracefully and take corrective action without stopping the excecution of program.

# What is an Exception?

An exception is an error that occurs during program execution and interrupts the normal flow of the program.Common exceptions include:

ZeroDivisionError: Dividing by zero
FileNotFoundError: File not Found.
ValueError:Invalid value.
TypeError:Invalid type.


```python
#Exception try,except block

try: 
    a=b
except:
    print("Ths variable has not been assigned")
```

```python
# Handling the exception using the exception object
try:
    a=b
except NameError as ex:
    print(ex) 
```

```python
try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter Denominaor other than zero")
except Exception as ex:
    print(ex)
```

```python
## try,except,else block
try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("This is not a valid number")
except ZeroDivisionError:
    print("Enter Denominaor other than zero")
except Exception as ex:
    print(ex)
else:
    print(f"The result is: {result} ")

''' 
if I want to execute some code if there is no exception then I can
use else block 
'''
```

```python
## try,except,else and finally block
try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("This is not a valid number") 
except ZeroDivisionError:
    print("Enter Denominaor other than zero")
else:
    print(f"The result is: {result} ")
finally:
    print("This block will be executed no matter what")

'''When we want to execute some code no matter what, 
we can use finally block.It is also executed even if 
there is an exception in the try block.
'''

'''What is the use of finally block in Python?
The realword use is if in case the user enter a wrong password
and the program is terminated, we can use finally block to close 
the database connection or file connection.'''
```
