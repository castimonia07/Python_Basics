### Joining Paths

```python
dir_name="folder"
file_name="file.txt"
full_path=os.path.join(os.getcwd(),dir_name,file_name)

print(full_path) # folder/file.txt
```

```python
## Check if a file exists

path='example1.txt'
if os.path.exists(path):
    print(f"This path'{path}' exists")
else:
    print(f"This path'{path}' does not exists")
```

```python
## Checking if a Path is a file or Directory

import os

path='example1.tst'
if os.path.isfile(path):
    print(f"This path '{path}' is a file")
elif os.path.isdir(path):
    print(f"This path '{path}' is a directory")
else:
    print(f"This path '{path}' is neither a file nor a directory")
```

```python
## Getting the absolute path of a file

relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
print(relative_path)
```
