# File Operations-Read and Write Files


```python
# Read a Whole File
with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
# Read a File Line by Line
with open('example.txt','r') as file:
    for line in file:
        print(line) # it will print the line with newline or line break is added automatically
        print(line.strip()) # it will remove the newline character from the end of the line
```

```python
# Writing a file(Overwriting)

with open('example.txt','w')as file:
    file.write('This is a new line\n') # it will overwrite the existing content of the file
    '''means file ko ye as an empty file consider krega and usme ki 
    har chiz hta ke ye new line add kr dega'''
```

```python
## Write a file(withouut Overwriting)
with open('example.txt','a') as file:
    file.write("Append operation taking place\n") # it will append the new line to the existing content of the file
```

```python
### writing a list of lines to a file

lines=['First line\n','Second line\n','Third line\n']
with open('example.txt','a')as file:
    file.writelines(lines) 
    # it will append the list of lines to the existing content of the file
```

```python
# Binary File Operations
data =b'\x00\x01\x02\x04'
with open('example.bin','wb')as file:
    file.write(data) # it will write the binary data to the file


with open('example.bin','rb')as file:
    content=file.read()
    print(content) # it will read the binary data from the file
```

```python
## Read the content from source test file and write to adestination text file

with open ('example.txt','r') as source_file:
    content= source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content) # it will write the content of source file to destination file
```

```python
## Read a text file and count the number of lines,words and character.

with open('example.txt','r') as file:
    content=file.read()
    characters=len(content)
    words=len(content.split())
    lines=len(content.splitlines())
    print(f"Number of characters:",{characters})
    print(f"Number of words:",{words})
    print(f"Number of lines:",{lines})   
```

# W+ mode
